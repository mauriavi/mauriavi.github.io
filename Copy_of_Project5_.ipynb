## Loading Data



### Function to fetch data


```python
def fetch_text(raw_url):

  """
  Fetches text content from a given URL, utilizing a local cache to prevent redundant downloads.

  The function first checks if the content from the `raw_url` is already present in a local cache
  directory (`cs_110_content/text_cache`). If cached, it returns the stored content.
  Otherwise, it sends an HTTP GET request to the URL, saves the fetched content to the cache,
  and then returns it. Error handling is included to catch potential issues during the fetch process.

  Parameters:
    raw_url (str): The URL of the text file to fetch.

  Returns:
    str: The fetched text content if successful.
    None: If an error occurs during fetching or the content cannot be retrieved.

  Notes:
  - The cache directory `cs_110_content/text_cache` is created if it does not already exist.
  - A SHA-1 hash of the URL is used to create a unique filename for caching.
  - A timeout of 10 seconds is applied to the HTTP GET request.
  - `requests.raise_for_status()` is used to raise an HTTPError for bad responses (4xx or 5xx).
  - The function prints a success or failure message to the console.
  """

  import requests
  from pathlib import Path
  import hashlib

  CACHE_DIR = Path("cs_110_content/text_cache")
  CACHE_DIR.mkdir(parents=True, exist_ok=True)

  def _url_to_filename(url):
    url_hash = hashlib.sha1(url.encode("utf-8")).hexdigest()[:12]
    return CACHE_DIR / f"{url_hash}.txt"

  cache_path = _url_to_filename(raw_url)

  SUCCESS_MSG = "✅ Text fetched."
  FAILURE_MSG = "❌ Failed to fetch text."
  try:
    if not cache_path.exists():
      response = requests.get(raw_url, timeout=10)
      response.raise_for_status()
      text_data = response.text
      cache_path.write_text(text_data, encoding="utf-8")
    print(SUCCESS_MSG)
    return cache_path.read_text(encoding="utf-8")

  except Exception as e:
    print(FAILURE_MSG)
    print(f"Error: {e}")
    return ""
```

```python
wuthering_heights_URL = "https://www.gutenberg.org/cache/epub/768/pg768.txt"

wuthering_heights_text = fetch_text(wuthering_heights_URL)
```

```python
import operator

def print_top_10_frequent_words(text):
    """
    Prints the top 10 most frequent words from a given text.

    This function first tokenizes and counts the frequency of each word in the input text
    (by calling `get_word_counts`). It then sorts these word counts in descending order
    and prints the top 10 most frequent words along with their respective counts.

    Parameters:
      text (str): The input text string to analyze.

    Returns:
      None: This function does not return any value; it prints the output directly to the console.

    Notes:
    - Word counting is case-insensitive (all words are converted to lowercase).
    - The definition of a "word" is determined by the `get_word_counts` function,
      which uses whitespace as a delimiter and includes punctuation attached to words.
    - If the text contains fewer than 10 unique words, all unique words will be printed.
    """
    word_counts = get_word_counts(text)
    sorted_word_counts = dict(sorted(word_counts.items(), key=operator.itemgetter(1), reverse=True))
    top_10_words = list(sorted_word_counts.items())[:10]  # Get the top 10 words and counts
    for word, count in top_10_words:
        print(f"{word}: {count}")

print_top_10_frequent_words(wuthering_heights_text)
```

```python
def word_tokenization_normalization(text):
    """
    Performs tokenization and normalization on the input text using spaCy.

    This function processes raw text by converting it to lowercase, tokenizing it,
    and then normalizing the tokens. Normalization includes removing stopwords,
    punctuation, numbers, and words shorter than three characters. The remaining
    words are lemmatized to their base form.

    Parameters:
      text (str): The input text string to be processed.

    Returns:
      list: A list of normalized and lemmatized words (strings) from the input text.

    Notes:
    - The text is first converted to lowercase.
    - Stopwords (common words like 'the', 'is'), punctuation, and numbers are removed.
    - Only words with a length greater than 2 characters after stripping whitespace are kept.
    - Words are returned in their base (lemmatized) form.
    """

    text = text.lower() # lowercase
    doc = nlp(text)     # loading text into model

    words_normalized = []
    for word in doc:
        if word.text != '\n' \
        and not word.is_stop \
        and not word.is_punct \
        and not word.like_num \
        and len(word.text.strip()) > 2:
            word_lemmatized = str(word.lemma_)
            words_normalized.append(word_lemmatized)

    return words_normalized
```

```python
def word_count(word_list):
    """
    Calculates the frequency of each word in a given list of words.

    This function iterates through a list of words, converts each word to lowercase,
    and counts its occurrences. It returns a dictionary where keys are the unique
    words (in lowercase) and values are their respective counts.

    Parameters:
      word_list (list): A list of strings, where each string is a word.

    Returns:
      dict: A dictionary where keys are words (str) and values are their counts (int).

    Notes:
    - All words are converted to lowercase before counting.
    - The function does not perform any tokenization or cleaning; it assumes the input
      `word_list` already contains individual words.
    """
    word_counts = {}
    for word in word_list:
      word = word.lower()
      if word in word_counts:
        word_counts[word] += 1
      else:
        word_counts[word] = 1
    return word_counts
```

```python
import operator

def print_top_15_frequent_words(word_counts):
    """
    Prints the top 15 most frequent words from a given word count dictionary.

    This function sorts the word count dictionary in descending order based on
    the word frequencies and then prints the top 15 words along with their counts.

    Parameters:
      word_counts (dict): A dictionary where keys are words (str) and values are their counts (int).

    Returns:
      None: This function does not return any value; it prints the output directly.

    Notes:
    - The function uses the `operator.itemgetter(1)` to sort the dictionary items by their values (counts).
    - It takes a slice of the sorted list to get only the top 15 words.
    - If the dictionary contains fewer than 15 unique words, all words will be printed.
    """
    sorted_word_counts = dict(sorted(word_counts.items(), key=operator.itemgetter(1), reverse=True))
    top_15_words = list(sorted_word_counts.items())[:15]  # Get the top 15 words and counts
    for word, count in top_15_words:
        print(f"{word}: {count}")
```

```python
import operator

def get_verb_counts(text):
    """
    Extracts, tokenizes, normalizes, and counts verbs from the input text using spaCy.

    This function processes raw text by converting it to lowercase, tokenizing it,
    and then normalizing the tokens. Normalization includes identifying verbs,
    removing stopwords, punctuation, numbers, and words shorter than three characters.
    The remaining verbs are lemmatized to their base form and their frequencies are counted.

    Parameters:
      text (str): The input text string to be processed.

    Returns:
      dict: A dictionary where keys are lemmatized verbs (str) and values are their counts (int).
    """
    # Assuming 'nlp' is already loaded globally by spacy.load('en_core_web_sm')
    # If not, it should be loaded here: nlp = spacy.load('en_core_web_sm')

    text = text.lower() # lowercase
    doc = nlp(text)     # loading text into model

    verb_counts = {}
    for token in doc:
        # Check if the token is a verb and meets other normalization criteria
        if token.pos_ == "VERB" \
        and not token.is_stop \
        and not token.is_punct \
        and not token.like_num \
        and len(token.text.strip()) > 2: # Ensure it's a verb and not a short, meaningless token
            lemma = token.lemma_
            verb_counts[lemma] = verb_counts.get(lemma, 0) + 1
    return verb_counts

# Get verb counts
verb_counts = get_verb_counts(wuthering_heights_text)

# Sort the verb counts and get the top 10
sorted_verb_counts = dict(sorted(verb_counts.items(), key=operator.itemgetter(1), reverse=True))
top_10_verbs = list(sorted_verb_counts.items())[:10]

print("Top 10 Most Frequent Verbs:")
for verb, count in top_10_verbs:
    print(f"{verb}: {count}")
```
